DICCIONARIO DE DATOS: COFFE QUALITY DATASET

1. Identidad y Trazabilidad

 - Unnamed: 0: Índice numérico residual del archivo original; no tiene valor analítico.
 
 - Species: Tipo de especie botánica (Arabica o Robusta).
 
 - Owner / Owner.1: Persona o entidad dueña del lote de café.

 - Country.of.Origin: País de procedencia del grano.

 - Farm.Name: Nombre de la finca productora.

 - Lot.Number: Código de identificación del lote específico.

 - Mill: Nombre del beneficio o planta donde se procesó el café.

 - ICO.Number: Registro internacional ante la International Coffee Organization.

 - Company: Empresa exportadora o comercializadora.

 - Region: Zona geográfica específica dentro del país de origen.

 - Producer: Nombre del productor o caficultor.

 - In.Country.Partner: Organización que coordina la certificación en el país origen (ej. Almacafé).

2. Logística y Certificación

 - Number.of.Bags: Cantidad total de sacos en el lote.

 - Bag.Weight: Peso de cada saco individual.

 - Harvest.Year: Periodo o año de la cosecha.

 - Grading.Date: Fecha en la que se realizó la evaluación técnica.

 - Expiration: Fecha límite de validez de la certificación.

 - Certification.Body: Entidad encargada de emitir el certificado de calidad.

 - Certification.Address / Contact: Datos de ubicación y contacto de la certificadora.

3. Atributos Sensoriales

 - Aroma: Fragancia del café en seco y su olor al contacto con agua caliente.

 - Flavor: Intensidad y calidad del sabor percibido.

 - Aftertaste: Calidad del sabor remanente tras la degustación.

 - Acidity: Nivel de brillantez y viveza del grano.

 - Body: Textura y sensación de peso en el paladar.

 - Balance: Equilibrio entre aroma, sabor, acidez y cuerpo.

 - Uniformity: Consistencia de sabor entre las distintas tazas de la muestra.

 - Clean.Cup: Ausencia de sabores defectuosos o "sucios".

 - Sweetness: Presencia de dulzor natural en el grano.

 - Cupper.Points: Calificación global otorgada por el catador.

 - Total.Cup.Points: Variable objetivo (Target). Puntaje final sobre 100 puntos.

4. Atributos Físicos y Defectos

 - Moisture: Grado de humedad del grano verde.

 - Color: Tono visual del café (Verde, Azulado, etc.).

 - Category.One.Defects: Conteo de defectos primarios (los más graves).

 - Category.Two.Defects: Conteo de defectos secundarios (leves).

 - Quakers: Granos que no se tuestan bien por falta de madurez.

5. Altitud

 - Altitude: Texto original con la información de altura (sin procesar).

 - unit_of_measurement: Unidad utilizada en el registro original (ft o m).

 - altitude_low_meters: Valor mínimo del rango de altitud en metros.

 - altitude_high_meters: Valor máximo del rango de altitud en metros.

 - altitude_mean_meters: Promedio de altitud calculado y normalizado

# 5. Limpieza y relleno de valores faltantes

```python
#Las columnas que eliminamos no aportan valor o tienen demasiados nulos son las siguientes:

cols_to_drop = [
    'Lot.Number', 'Farm.Name', 'Mill', 'Producer', 'Company', 
    'ICO.Number', 'Owner.1',
    'Quakers', 'Expiration', 'Certification.Body', 'Unnamed: 0',    
    'Harvest.Year', 
    #Aclaración: Si bien 'Harvest.Year' podría ser interesante, tiene demasiados nulos, 
    #por lo que decidimos utilizar Grading.Date como fecha de referencia aproximada de cosecha.
    'Certification.Address', 'Certification.Contact', 'Altitude']
df_merged_limpio = df_merged.drop(columns=cols_to_drop, errors='ignore')
```

En cuanto a la columna 'Owner', decidimos rellenar los valores nulos con 'Desconocido' para mantener la información de que no se conoce el propietario, en lugar de eliminar la columna o dejarla con valores nulos.

```python
df_merged_limpio['Owner'] = df_merged_limpio['Owner'].fillna('Desconocido')
```

Para la columna 'Country.of.Origin', decidimos rellenar el unico valor nulo con 'United States' ya que identificamos que el café con ese In.Country.Partner provenía de ese país.

```python
df_merged_limpio.loc[df_merged_limpio['Country.of.Origin'].isnull(), 'Country.of.Origin'] = 'United States'
```

Ahora, para las columnas categóricas restantes, vamos a rellenar los valores nulos con la moda de cada grupo definido por 'Country.of.Origin' y 'Species'.

```python
cols_moda = ['Color', 'Variety', 'Processing.Method']

for col in cols_moda:
    moda_por_grupo = df_merged_limpio.groupby(['Country.of.Origin', 'Species'])[col].transform(
        lambda x: x.mode()[0] if not x.mode().empty else "No Especificado"
    )
    df_merged_limpio[col] = df_merged_limpio[col].fillna(moda_por_grupo)

# Para la Región, simplemente rellenamos con "No Especificado" sin buscar la moda
df_merged_limpio['Region'] = df_merged_limpio['Region'].fillna("No Especificado")

# Verificación de seguridad
print("Nulos por columna:")
print(df_merged_limpio[cols_moda + ['Region']].isnull().sum())
```

# 6. Limpiar y estandarizar las unidades de altitud

Finalmente, estandarizamos las unidades de altitud a metros, asumiendo que si la unidad es pies, se debe convertir a metros.

```python
def estandarizar_unidades(row):
    unidad = str(row['unit_of_measurement']).lower().strip()
    
    if 'ft' in unidad or 'feet' in unidad:
        cols = ['altitude_low_meters', 'altitude_high_meters', 'altitude_mean_meters']
        for col in cols:
            if pd.notnull(row[col]):
                row[col] = row[col] * 0.3048
    return row
df_merged_limpio = df_merged_limpio.apply(estandarizar_unidades, axis=1)
```

Eliminamos las columnas de altitud Low y High porque ya tenemos la columna de altitud media, que es la que utilizaremos para el análisis, y así simplificamos el DataFrame.

```python
df_merged_limpio= df_merged_limpio.drop(columns=['altitude_low_meters', 'altitude_high_meters', 'unit_of_measurement'], errors='ignore')
print(df_merged_limpio.isnull().sum().sort_values(ascending=False).head(50))
```

Finalmente rellenamos los valores nulos de la columna 'altitude_mean_meters' utilizando la mediana de altitud para cada combinación de 'Country.of.Origin' y 'Species'.

```python
# Paso A: Rellenar usando la mediana del País Y la Especie
df_merged_limpio['altitude_mean_meters'] = df_merged_limpio['altitude_mean_meters'].fillna(
    df_merged_limpio.groupby(['Country.of.Origin', 'Species'])['altitude_mean_meters'].transform('median')
)
# Paso B: Si aún quedan nulos (países con un solo registro nulo), usar la mediana de la Especie
df_merged_limpio['altitude_mean_meters'] = df_merged_limpio['altitude_mean_meters'].fillna(
    df_merged_limpio.groupby('Species')['altitude_mean_meters'].transform('median')
)
print(f"Nulos restantes en altitud: {df_merged_limpio['altitude_mean_meters'].isnull().sum()}")
```

Agregamos una nueva columna que categorice la altitud en rangos, diferenciando entre Arabica y Robusta, para facilitar el análisis posterior.

```python
def categorizar_altitud_pro(row):
    metros = row['altitude_mean_meters']
    especie = row['Species']
    
    if especie == 'Arabica':
        if metros < 1000: return 'Baja (Arabica)'
        if 1000 <= metros < 1500: return 'Media (Arabica)'
        if 1500 <= metros < 1800: return 'Alta (Arabica)'
        return 'Premium (Arabica)'
    
    else:
        if metros < 400: return 'Baja (Robusta)'
        if 400 <= metros < 700: return 'Media (Robusta)'
        return 'Alta (Robusta)'

df_merged_limpio['categoria_altitud'] = df_merged_limpio.apply(categorizar_altitud_pro, axis=1)
```
